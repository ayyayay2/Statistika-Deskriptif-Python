```python
import pandas as pd

df = pd.read_excel(
    'ayya.xlsx',
    sheet_name='Sheet1',
    header=6
)

# Mengambil hanya data mahasiswa
df = df[df['ID'].notna()].copy()

# Mengambil 7 kolom utama
df = df.iloc[:, :8]

df.head()
```

```python
print("Jumlah mahasiswa:", len(df))
print(df.columns.tolist())
```

```python
final_score = pd.to_numeric(
    df['Final Score'],
    errors='coerce'
)
```

```python
final_score.describe()
```

```python
mean = final_score.mean()
median = final_score.median()
modus = final_score.mode().iloc[0]
minimum = final_score.min()
maximum = final_score.max()

Q1 = final_score.quantile(0.25)
Q2 = final_score.quantile(0.50)
Q3 = final_score.quantile(0.75)

IQR = Q3 - Q1

variance = final_score.var()
std = final_score.std()

standard_error = std / (len(final_score) ** 0.5)

skewness = final_score.skew()
kurtosis = final_score.kurt()

print("Mean:", mean)
print("Median:", median)
print("Modus:", modus)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
print("Variance:", variance)
print("Standard Deviation:", std)
print("Standard Error:", standard_error)
print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
```

```python
frekuensi_grade = df['Grade'].value_counts()

print(frekuensi_grade)
```

```python
urutan_grade = ['A', 'AB', 'B', 'BC', 'C', 'F']

frekuensi_grade = (
    df['Grade']
    .value_counts()
    .reindex(urutan_grade, fill_value=0)
)

print(frekuensi_grade)
```

```python
tabel_frekuensi = pd.DataFrame({
    'Grade': frekuensi_grade.index,
    'Frequency': frekuensi_grade.values
})

tabel_frekuensi
```

```python
tabel_frekuensi.plot(
    x='Grade',
    y='Frequency',
    kind='line',
    marker='o',
    title='Line Chart Frekuensi Grade'
)
```

```python
tabel_frekuensi.plot(
    x='Grade',
    y='Frequency',
    kind='bar',
    title='Bar Chart Frekuensi Grade'
)
```

```python
tabel_frekuensi.set_index('Grade')['Frequency'].plot(
    kind='pie',
    autopct='%1.1f%%',
    title='Pie Chart Persentase Grade'
)
```

## Interpretasi Hasil Analisis

Berdasarkan hasil statistika deskriptif terhadap 31 data nilai mahasiswa, diperoleh nilai mean sebesar 76,75 dan median sebesar 79,33. Nilai median lebih tinggi dibandingkan mean, yang menunjukkan adanya kecenderungan distribusi data miring ke kiri.

Nilai modus sebesar 77,33 menunjukkan bahwa nilai tersebut merupakan nilai yang paling sering muncul dalam data.

Nilai minimum adalah 46 dan nilai maksimum adalah 94, sehingga rentang nilai mahasiswa cukup luas.

Nilai Q1 sebesar 68,83, Q2 sebesar 79,33, dan Q3 sebesar 87,50. Nilai IQR sebesar 18,67 menunjukkan penyebaran 50% data yang berada di bagian tengah.

Standard deviation sebesar 13,56. Berdasarkan acuan pada materi, nilai tersebut berada pada rentang 10 sampai 15 sehingga variasi nilai termasuk wajar untuk data dengan skala 0 sampai 100.

Nilai skewness sebesar -0,70 menunjukkan adanya kecenderungan kemencengan negatif, tetapi masih berada dalam rentang -1 sampai +1 yang pada materi digunakan sebagai acuan bahwa distribusi masih cukup dekat dengan simetris.

Nilai kurtosis sebesar -0,49. Karena Pandas menggunakan excess kurtosis, nilai yang mendekati 0 menunjukkan karakteristik distribusi yang mendekati normal secara deskriptif.

```python
batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

print("Batas bawah:", batas_bawah)
print("Batas atas:", batas_atas)

outlier = final_score[
    (final_score < batas_bawah) |
    (final_score > batas_atas)
]

print("Jumlah calon outlier:", len(outlier))
print("Data calon outlier:")
print(outlier)
```
